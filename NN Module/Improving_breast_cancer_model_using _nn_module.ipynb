```python
import pandas as pd 
import numpy as np 
import torch 
import torch.nn as nn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.metrics import accuracy_score
import torch.nn as nn
import torch.optim as optim 
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.columns
```

```python
df.shape
```

```python
df.drop(columns=['id','Unnamed: 32'],inplace=True)
df 
```

```python
X_train,X_test,y_train,y_test=train_test_split(df.drop(columns=['diagnosis']),df['diagnosis'],test_size=0.2,random_state=23,shuffle=True)
```

```python
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
X_train
```

```python
label=LabelEncoder()
y_train=label.fit_transform(y_train)
y_test=label.transform(y_test)
```

```python
y_train
```

```python
X_train_ten=torch.tensor(X_train,requires_grad=True,dtype=torch.float32)
X_test_ten=torch.tensor(X_test,requires_grad=True,dtype=torch.float32)
y_train_ten=torch.tensor(y_train,dtype=torch.float32)
y_test_ten=torch.tensor(y_test,dtype=torch.float32)
```

```python
X_train_ten.dtype 
```

```python
X_train_ten
```

```python
class MyNeuralNetwork(nn.Module):
 
    def __init__(self,feature_size):
        super().__init__()
        self.linear=nn.Linear(feature_size,1)
        self.sigmoid=nn.Sigmoid()
 
    def forward(self,X): 
        out=self.linear(X)
        out=self.sigmoid(out)
        return out
        
        
    def loss(self,y_pred,y): 
        bce=nn.BCELoss()
        los=bce(y_pred,y)
        return los
    
    def backpropagation(self,loss): 
        loss.backward()
        

    
        
```

```python
model=MyNeuralNetwork(X_train_ten.shape[1])
```

```python
lr=0.001
epochs=75
```

```python
model.linear.weight
```

```python
model.linear.bias
```

```python
optimizer=optim.Adam(model.parameters(),lr=lr)
optimizer
```

```python
for epoch in range(epochs): 
    y_pred=model(X_train_ten) 
    loss=model.loss(y_pred,y_train_ten.reshape(-1,1))
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    print(f'Loss for Epoch {epoch+1} Loss {loss}')
    
```

```python
X_test_ten.shape
```

```python
model.linear.weight.reshape(-1,1).shape 
```

```python
y_pred_test = torch.sigmoid(X_test_ten @ model.linear.weight.reshape(-1,1) + model.linear.bias)
y_pred_test
```

```python
y_pred_test.shape 
```

```python
y_pred_class = (y_pred_test >= 0.5).double()
y_pred_class
```

```python
accuracy_score(
    y_test_ten.detach().numpy(),
    y_pred_class.detach().numpy()
)
```
